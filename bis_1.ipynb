```python
import random

# 1. Setup Parameters & Mock Data (50 Items)
random.seed(42)  # Seed for reproducible results
Items = [{"Weight": random.randint(50, 150), "Revenue": random.randint(100, 500)} for _ in range(50)]

MaxCapacity = 3500
PopSize = 50
Generations = 100
CrossoverRate = 0.80
MutationRate = 0.02

# Helper function for Tournament Selection
def RunTournament(population, fitness_scores, size=3):
    participants = random.sample(list(zip(population, fitness_scores)), size)
    participants.sort(key=lambda x: x[1], reverse=True)  # Sort by fitness descending
    return participants[0][0]  # Return the best individual

# 1. Initialization
Population = [[random.randint(0, 1) for _ in range(50)] for _ in range(PopSize)]
BestSolution = None
BestRevenue = float('-inf')

# 2. Evolutionary Loop
for gen in range(1, Generations + 1):
    FitnessScores = []

    # Step A: Evaluation
    for individual in Population:
        TotalWeight = sum(ind * item["Weight"] for ind, item in zip(individual, Items))
        TotalRevenue = sum(ind * item["Revenue"] for ind, item in zip(individual, Items))

        # Constraint Penalty
        if TotalWeight > MaxCapacity:
            Fitness = 0
        else:
            Fitness = TotalRevenue

        FitnessScores.append(Fitness)

        # Track global elite solution
        if Fitness > BestRevenue:
            BestRevenue = Fitness
            BestSolution = list(individual)

    # Step B: Reproduction
    NextGeneration = []
    while len(NextGeneration) < PopSize:
        # Selection
        Parent1 = RunTournament(Population, FitnessScores, size=3)
        Parent2 = RunTournament(Population, FitnessScores, size=3)

        # Crossover
        if random.random() < CrossoverRate:
            CrossoverPoint = random.randint(1, 48)
            Child1 = Parent1[:CrossoverPoint] + Parent2[CrossoverPoint:]
            Child2 = Parent2[:CrossoverPoint] + Parent1[CrossoverPoint:]
        else:
            Child1 = list(Parent1)
            Child2 = list(Parent2)

        # Mutation
        for child in [Child1, Child2]:
            for i in range(50):
                if random.random() < MutationRate:
                    child[i] = 1 - child[i]  # Invert bit

            if len(NextGeneration) < PopSize:
                NextGeneration.append(child)

    # Step C: Generation Shift
    Population = NextGeneration

    # Shorter log for tracking progress
    if gen in [1, 20, 50, 100]:
        print(f"Generation {gen:03d} -> Best Revenue: {BestRevenue}")

# Calculate final weight of the elite solution
FinalWeight = sum(ind * item["Weight"] for ind, item in zip(BestSolution, Items))

print("\n--- FINAL RESULTS ---")
print(f"Optimal Cargo Array : {''.join(map(str, BestSolution))}")
print(f"Total Weight Payload : {FinalWeight} kg")
print(f"Maximum Best Revenue : {BestRevenue} Units")
```
